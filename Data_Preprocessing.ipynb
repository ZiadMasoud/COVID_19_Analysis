```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('File 1.xlsx', sheet_name='covid_19_data')
df.head()
```

```python
# Check for null values and basic info
df.info(), df.isnull().sum(), df.describe(), df.shape, df.dtypes, df.columns, df.head(), df.value_counts()
```

# 🧼 Data Cleaning Summary
✅ Key Points:
* Total rows: 205,951
* Columns: 8
* Column Province/State has ~55,000 missing values → that’s expected for countries without provinces (e.g., Italy, Iran).
* All other columns are clean and complete.
* Dates (ObservationDate) are stored as strings, so we’ll convert them to datetime for proper analysis.

```python
# Convert date column to datetime
df['ObservationDate'] = pd.to_datetime(df['ObservationDate'])

# Group by Date and Country, summing confirmed, deaths, and recovered
grouped_df = df.groupby(['ObservationDate', 'Country/Region'], as_index=False)[['Confirmed', 'Deaths', 'Recovered']].sum()

# Save the cleaned and grouped data for Tableau
cleaned_path = "covid_cleaned_for_tableau.csv"
grouped_df.to_csv(cleaned_path, index=False)

# Show the first few rows of the cleaned dataset
print(grouped_df.head())
```
